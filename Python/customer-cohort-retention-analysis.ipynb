```python
import pandas as pd

df = pd.read_excel("E:/Shubhangi/Data analysis projects/Customer Cohort Retention Analysis/Dataset/Online Retail.xlsx")

df.head()
```

```python
df = df.dropna(subset=['CustomerID'])
```

```python
df = df[df['Quantity'] > 0]
```

```python
df = df[df['UnitPrice'] > 0]
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```
